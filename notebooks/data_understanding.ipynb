# EduPredict: Student Academic Performance - Data Understanding

This notebook focuses on the initial exploration of the synthetic dataset for predicting student academic performance.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load dataset
df = pd.read_csv('../dataset/raw_data/student_data.csv')
print("Dataset loaded successfully.")
```

## 1. Basic Information
Let's inspect the data types and null values.

```python
df.info()
```

```python
df.head()
```

## 2. Descriptive Statistics
Check the distribution of numerical features.

```python
df.describe()
```

## 3. Data Integrity
Check for missing values and duplicates.

```python
print("Null values:\n", df.isnull().sum())
print("\nDuplicate records:", df.duplicated().sum())
```

## 4. Categorical Breakdown
Check value counts for gender and performance labels.

```python
print("Gender distribution:\n", df['gender'].value_counts())
print("\nPerformance label distribution:\n", df['performance_label'].value_counts())
print("\nFinal grade distribution:\n", df['final_grade'].value_counts())
```

## 5. Correlation Analysis
Visualize relationships between numerical features.

```python
numeric_df = df.select_dtypes(include=[np.number])
plt.figure(figsize=(10, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Matrix')
plt.show()
```
